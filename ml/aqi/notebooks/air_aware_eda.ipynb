# LifeGuard AI — Module 4: Air Aware (Chronological Split, Lags & Forecast Audit)

This notebook audits the 12-station daily air quality series (`2013-03-01` → `2017-02-28`, `17,441` consecutive `(t -> t+1)` day pairs), enforces strict chronological splitting (**no random split**), evaluates expanding-window `TimeSeriesSplit` CV against a Persistence baseline (`AQI_{t+1} = AQI_t`), and inspects winter vs non-winter residuals on the untouched future test period.

```python
import json
from pathlib import Path
import pandas as pd

REPO_ROOT = Path("../../..").resolve()
META_PATH = REPO_ROOT / "ml" / "aqi" / "artifacts" / "metadata.json"

entries = json.loads(META_PATH.read_text(encoding="utf-8"))
print(f"Loaded {len(entries)} Air Aware model reports.")
```

## 1. Chronological Split & Causal Feature Verification

- **Zero Future Leakage**: Trailing 3-day / 7-day AQI averages and 1-day lag are computed strictly per station over historical indices (`<= t`).
- **Strict Time Split**: Training/validation covers `2013-03-01 -> 2016-05-11` (`13,947` rows); the final 20% future period (`2016-05-12 -> 2017-02-27`, `3,494` rows) is held out completely untouched.

```python
for entry in entries:
    print(f"=== {entry['sub_model'].upper()} TRACK ===")
    print("Splits:", entry["splits"])
    display(pd.DataFrame(entry["comparison_table"]))
```
